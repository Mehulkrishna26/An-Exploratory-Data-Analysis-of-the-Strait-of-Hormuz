Descriptive Analysis

```python
Chart 1 — Countplot: Distribution of trade tiers



plt.figure(figsize=(8, 5))

sns.countplot(data=df, x='trade_tier',
              order=df['trade_tier'].value_counts().index)

plt.title('Distribution of Trade Tiers')
plt.xlabel('Trade Tier')
plt.ylabel('Number of Records')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```


**Countplot:** Distribution of trade tiers
This shows how many records belong to each trade tier.

Identify the most frequently represented trade tier.



Compare the distribution of Privileged, Taxed, and any other tiers present.



This helps assess how trade records are distributed across different toll or trade categories

Chart 2 — Histogram: Total transit cost

```python



plt.figure(figsize=(8, 5))

sns.histplot(data=df, x='total_transit_cost_usd',
             bins=20, kde=True)

plt.title('Distribution of Total Transit Costs')
plt.xlabel('Total Transit Cost (USD)')
plt.ylabel('Number of Records')
plt.ticklabel_format(style='plain', axis='x')
plt.tight_layout()
plt.show()
```

Histogram: Total transit cost :Identify the typical range of transit costs.



A long right tail indicates that some records have much higher costs than others.



This helps identify cost concentration and observations that deserve further investigation.


```python
Chart 3 — Boxplot: Days delayed
```

```python

plt.figure(figsize=(8, 4))

sns.boxplot(data=df, x='days_delayed')

plt.title('Distribution of Shipping Delays')
plt.xlabel('Days Delayed')
plt.tight_layout()
plt.show()
```

Boxplot: Days delayed : Examine the median delay and spread of delays.



Potential outliers may represent unusually delayed shipments.



These records can be investigated to understand their relationship with rerouting, transit status, and additional costs.

```python
Chart 4 — Barplot: Average transit cost by trade tier
```

```python

plt.figure(figsize=(9, 5))

sns.barplot(data=df, x='trade_tier',
            y='total_transit_cost_usd',
            errorbar=None)

plt.title('Average Transit Cost by Trade Tier')
plt.xlabel('Trade Tier')
plt.ylabel('Average Transit Cost (USD)')
plt.ticklabel_format(style='plain', axis='y')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

Barplot: Average transit cost by trade tier :
Compare average transit costs across trade tiers.



Determine which tier has the highest average recorded cost.



This can help explore differences in the cost structure between tiers. Higher costs do not necessarily mean higher tolls, because insurance, fuel, and rerouting expenses may also contribute.

**Scatterplot**: Cargo value versus transit cost


```python

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x='estimated_cargo_value_usd',
    y='total_transit_cost_usd',
    hue='rerouted',
    alpha=0.7
)

plt.title('Cargo Value vs Total Transit Cost')
plt.xlabel('Estimated Cargo Value (USD)')
plt.ylabel('Total Transit Cost (USD)')
plt.ticklabel_format(style='plain', axis='both')
plt.legend(title='Rerouted')
plt.tight_layout()
plt.show()
```

**Scatterplot: Cargo value versus transit cost**
Check whether higher-value cargo is associated with higher transit costs.



Compare rerouted and non-rerouted records.



Look for unusual high-cost observations that may warrant further analysis.


**Chart 6 — Correlation heatmap**

```python

numeric_cols = [
    'toll_usd',
    'insurance_cost_usd',
    'days_delayed',
    'extra_fuel_tonnes',
    'reroute_penalty_usd',
    'total_transit_cost_usd',
    'estimated_cargo_value_usd',
    'total_asset_value_at_risk_usd',
    'inflation_premium_per_unit'
]

corr = df[numeric_cols].corr()

plt.figure(figsize=(11, 8))

sns.heatmap(corr, annot=True, cmap='coolwarm',
            fmt='.2f', vmin=-1, vmax=1, center=0)

plt.title('Correlation Heatmap of Maritime Cost and Risk Variables')
plt.tight_layout()
plt.show()
```

**Chart 7 — Pairplot: Cost, delay, and cargo value**

```python

features = [
    'total_transit_cost_usd',
    'insurance_cost_usd',
    'days_delayed',
    'estimated_cargo_value_usd'
]

pair_df = df[features].dropna()

sns.pairplot(pair_df, corner=True, diag_kind='hist')
plt.show()
```

**Pairplot: Cost, delay, and cargo value**
Examine relationships among cargo value, insurance, total cost, and delays simultaneously.



Identify clusters or unusual observations.



These patterns can help determine which variables merit closer investigation.

Chart 8 — Monthly trend: Total transit costs


```python

df['date'] = pd.to_datetime(df['date'], errors='coerce')

monthly_cost = (
    df.dropna(subset=['date'])
      .assign(month=lambda x: x['date'].dt.to_period('M'))
      .groupby('month', as_index=False)['total_transit_cost_usd']
      .sum()
)

monthly_cost['month'] = monthly_cost['month'].astype(str)

plt.figure(figsize=(11, 5))

sns.lineplot(data=monthly_cost, x='month',
             y='total_transit_cost_usd', marker='o')

plt.title('Monthly Total Maritime Transit Costs')
plt.xlabel('Month')
plt.ylabel('Total Transit Cost (USD)')
plt.xticks(rotation=45)
plt.ticklabel_format(style='plain', axis='y')
plt.tight_layout()
plt.show()
```

**Monthly trend:** Total transit costs
Identify months with the highest and lowest recorded total transit costs.



Investigate whether monthly changes coincide with changes in traffic, rerouting, or insurance expenses.



A rise in total costs might reflect more records as well as higher costs per vessel, so consider comparing monthly averages too.
